```python
import zipfile

# Define the downloaded zip file name
zip_file = 'house-prices-advanced-regression-techniques.zip'

# Extract the zip file into a folder named 'house-prices-data'
with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall('house-prices-data')

# Check the extracted files
print(os.listdir('house-prices-data'))
```
